## 6. Univariate Analysis

#### Numeric Feature(distribution)

```python
numeric_cols = ds.select_dtypes(include=['int', 'float']).columns

for col in numeric_cols:
    plt.figure(figsize=(5,3))
    sns.histplot(ds[col], color='#8B0000',kde=True)
    plt.title(f"Distribution of {col}")
    plt.show()
```

#### Outlier detection

```python
for col in numeric_cols:
    plt.figure(figsize=(5,3))
    sns.boxplot(x=ds[col],color='#8B0000')
    plt.title(f"Boxplot of {col}")
    plt.show()
```

## 7. Categorical Feature Analysis

```python
categorical_cols = ds.select_dtypes(include=['object']).columns

for col in categorical_cols:
    plt.figure(figsize=(6,3))
    sns.countplot(x=ds[col],color='#8B0000')
    plt.title(f"Count Plot of {col}")
    plt.show()
```

## 8. Bivariate Analysis

```python
target = "classification"  # Change this to your target column

for col in numeric_cols:
    plt.figure(figsize=(5,3))
    sns.boxplot(x=ds[target], y=ds[col],color='#8B0000')
    plt.title(f"{col} vs {target}")
    plt.show()
```

```python
for col in categorical_cols:
    plt.figure(figsize=(6,3))
    sns.countplot(x=ds[col], hue=ds[target],palette='dark:#8B0000')
    plt.title(f"{col} vs {target}")
    plt.show()
```

## 9. Correlation Analysis

```python
corr = ds.select_dtypes(include=['int64', 'float64']).corr()
corr
```

```python
plt.figure(figsize=(12,6))
sns.heatmap(corr, annot=True, cmap="Reds")
plt.show()
```
